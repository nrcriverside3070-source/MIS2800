## 4. Data Types & Mixed Lists

Every value in Python has a data type. `type()` tells you what you're working with.

```python
print(type(15))
print(type("hi"))
print(type(3.75))
print(type(True))

record = ["Alex", 21, 3.75, True]
print(record)
for item in record:
    print(item, "->", type(item))
```

**Note:** you haven't been taught loops yet, the `for item in record` line above is just there to show you every type in the list at once. You won't be asked to write a loop yourself this semester.

**A deeper example: checking types by index, without a loop.** This is the way you'll actually do it on HW1. Run the cell below.

## 5. Boolean Logic

A boolean only ever holds `True` or `False`. Comparison operators produce a boolean.

```python
age = 20
has_id = True

print(age >= 18)
print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
```

**FILL BELOW CODE CELL ALONG WITH VIDEO. A deeper example: combining several conditions at once.**

```python
# TODO: Fill code example from lecture here
gpa = 3.6
credit_hours = 65
has_hold = False

eligible_for_scholarship = gpa >= 3.5 and credit_hours >= 60 and not has_hold
print(eligible_for_scholarship)

eligible_for_registration = credit_hours >= 12 or not has_hold
print(eligible_for_registration)
```

Notice `not has_hold` flips `False` to `True`, since the student does *not* have a hold. Combining `and`/`or`/`not` like this is exactly how you'll filter and select data later this semester.

## 6. Dictionaries

A dictionary stores labeled values, a key paired with a value, instead of just position.

```python
student = {"name": "Alex", "major": "Business Analytics", "year": 3}

print(student["major"])

student["gpa"] = 3.5
student["year"] = 4

print(student.keys())
print(student.values())
print(student)
```
