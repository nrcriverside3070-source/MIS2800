```python
#this code imports sqlite3 and prints a statement stating it has been im
import sqlite3
print("sqlite3 imported")
```

```python
# This creates a brand-new, empty database file named 'AdCampaigns2026.d
conn = sqlite3.connect('AdCampaigns2026.db')
print("database created")
```

```python
%%sql sqlite:///AdCampaigns2026.db
CREATE TABLE Marketing_Brands (
  brand_id INTEGER PRIMARY KEY,
  brand_name TEXT,
  brand_type TEXT,
  brand_budget INTEGER
) STRICT;
```

```python
%%sql sqlite:///AdCampaigns2026.db
INSERT INTO Marketing_Brands (brand_id, brand_name, brand_type, brand_budget)
VALUES
  (1, 'Doritos', 'Snacks', 7000000),
  (2, 'AWS', 'Technology', 15000000),
  (3, 'Pepsi', 'Beverages', 2000000),
  (4, 'He_Gets_Us', 'Religious', 1000000),
  (5, 'Dr_Pepper', 'Beverages', 5000000),
  (6, 'IBM', 'Technology', 100000),
  (7, 'Google', 'Technology', 8000000);
```

```python
%%sql sqlite:///AdCampaigns2026.db
-- Instructions: Fix the code to select all the columns from the Marketing_Brands table.
SELECT * FROM Marketing_Brands;
```

```python
%%sql sqlite:///AdCampaigns2026.db
-- Instructions: Fix the code to display only the brand names and brand types.
SELECT brand_name, brand_type FROM Marketing_Brands;
```

```python
%%sql sqlite:///AdCampaigns2026.db
-- Instructions: Fix the code to find the brand types that are Religious
SELECT * FROM Marketing_Brands WHERE brand_type = 'Religious';
```

```python
%%sql sqlite:///AdCampaigns2026.db
-- Instructions: Fix the code to display the brand names that have a budget.
SELECT brand_name, brand_budget FROM Marketing_Brands;
```

```python
%%sql sqlite:///AdCampaigns2026.db
SELECT * FROM Marketing_Brands WHERE brand_budget > 7000000;
```

```python
%%sql sqlite:///AdCampaigns2026.db
--Instructions: Find the brands where the brand type is Snacks or Beverage
SELECT * FROM Marketing_Brands WHERE brand_type = 'Snacks' OR brand_type = 'Beverages';
```

```python
%%sql sqlite:///AdCampaigns2026.db
-- Instructions: Sort the table by brand budget in DESC order. Display all columns.
SELECT * FROM Marketing_Brands ORDER BY brand_budget DESC;
```

```python
%%sql sqlite:///AdCampaigns2026.db
-- Using SQLite3 (and all its magic) to create a table showing all the columns in ascending order
SELECT * FROM Marketing_Brands ORDER BY brand_budget ASC;
```
